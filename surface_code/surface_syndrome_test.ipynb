```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit import transpile
```

```python
z_stabilizers = [
    [0, 1, 3, 4],
    [1, 2, 4, 5],
    [3, 4, 6, 7],
    [4, 5, 7, 8],
]
```

```python
data = QuantumRegister(9, "data")
anc = QuantumRegister(4, "anc")
z_syn = ClassicalRegister(4, "z_syn")

qc = QuantumCircuit(data, anc, z_syn)
```

```python
def measure_z_stabilizer(qc, data, anc, z_syn, stabilizer, ancilla, syn_bit):
    for q in stabilizer:
        qc.cx(data[q], anc[ancilla])

    qc.measure(anc[ancilla], z_syn[syn_bit])
```

```python
# 1. Prepare/error
qc.x(data[6])

# 2. Build all syndrome measurements
for i, stabilizer in enumerate(z_stabilizers):
    measure_z_stabilizer(
        qc, data, anc, z_syn,
        stabilizer, i, i
    )

# 3. Run only after everything is built
sim = AerSimulator()
compiled = transpile(qc, sim)
result = sim.run(compiled, shots=1024).result()

counts = result.get_counts()
print(counts)
```

```python
print(qc.draw())
```

```python
data = QuantumRegister(9, "data")
anc = QuantumRegister(4, "anc")
x_syn = ClassicalRegister(4, "x_syn")


qc = QuantumCircuit(data, anc, x_syn)
```

```python
def measure_x_stabilizer(qc, data, anc, x_syn,
                         stabilizer, ancilla, syn_bit):

    qc.h(anc[ancilla])

    for q in stabilizer:
        qc.cx(anc[ancilla], data[q])

    qc.h(anc[ancilla])

    qc.measure(anc[ancilla], x_syn[syn_bit])
```

```python
x_stabilizers = [
    [0, 1, 3, 4],
    [1, 2, 4, 5],
    [3, 4, 6, 7],
    [4, 5, 7, 8],
]
```

```python
for q in range(9):
    qc.h(data[q])

# 1. Prepare/error
for j in range(9):
   data = QuantumRegister(9, "data")
   anc = QuantumRegister(4, "anc")
   x_syn = ClassicalRegister(4, "x_syn")


   qc = QuantumCircuit(data, anc, x_syn)
   for q in range(9):
      qc.h(data[q])
   qc.z(data[j])

   # 2. Build all syndrome measurements
   for i, stabilizer in enumerate(x_stabilizers):
      measure_x_stabilizer(
        qc, data, anc, x_syn,
        stabilizer, i, i
      )

    # 3. Run only after everything is built
   sim = AerSimulator()
   compiled = transpile(qc, sim)
   result = sim.run(compiled, shots=1024).result()

   counts = result.get_counts()
   print(counts)
```
