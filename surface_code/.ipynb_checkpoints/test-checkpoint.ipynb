```python
import numpy as np
from collections import deque
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# ============================================================
# 1. Complete Unified Decoder Graphs Configuration
# ============================================================

# The exact stabilizers from your notebook definitions
z_stabilizers_map = {
    0: [5, 6],
    1: [0, 1, 4, 5],
    2: [3, 4, 7, 8],
    3: [2, 3]
}

x_stabilizers_map = {
    0: [0, 1],
    1: [1, 2, 3, 4],
    2: [4, 5, 6, 7],
    3: [7, 8]
}

# GRAPH A: Graph used for decoding X errors (Bit Flips) via Z-Stabilizers
graph_z = {
    "Z0": [("Z1", 5), ("B_bottom", 6)],
    "Z1": [("B_top", 0), ("B_top", 1), ("Z2", 4), ("Z0", 5)],
    "Z2": [("Z3", 3), ("Z1", 4), ("B_bottom", 7), ("B_bottom", 8)],
    "Z3": [("B_top", 2), ("Z2", 3)],
    "B_top": [("Z1", 0), ("Z1", 1), ("Z3", 2)],
    "B_bottom": [("Z0", 6), ("Z2", 7), ("Z2", 8)],
}

boundary_connections_z = {
    "B_top": {"Z1": [0, 1], "Z3": [2]},
    "B_bottom": {"Z0": [6], "Z2": [7, 8]},
}

# GRAPH B: Graph used for decoding Z errors (Phase Flips) via X-Stabilizers
graph_x = {
    "X0": [("X1", 1), ("B_left", 0)],
    "X1": [("X0", 1), ("X2", 4), ("B_right", 2), ("X3", 3)],
    "X2": [("X1", 4), ("X3", 7), ("B_left", 5), ("B_left", 6)],
    "X3": [("X1", 3), ("X2", 7), ("B_right", 8)],
    "B_left": [("X0", 0), ("X2", 5), ("X2", 6)],
    "B_right": [("X1", 2), ("X3", 8)]
}

boundary_connections_x = {
    "B_left": {"X0": [0], "X2": [5, 6]},
    "B_right": {"X1": [2], "X3": [8]}
}

# ============================================================
# 2. Graph Search Helper Methods (Your MWPM Backend)
# ============================================================

def stabilizer_distance(graph, start, goal):
    queue = deque([(start, 0)])
    visited = {start}
    while queue:
        node, distance = queue.popleft()
        if node == goal: 
            return distance
        for neighbor, _ in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, distance + 1))
    return float("inf")

def boundary_distance(graph, boundary_connections, start, boundary):
    if boundary not in boundary_connections: 
        return float("inf")
    best = float("inf")
    for stabilizer in boundary_connections[boundary]:
        dist = stabilizer_distance(graph, start, stabilizer)
        if dist != float("inf"):
            best = min(best, dist + 1)
    return best

def boundary_aware_matching(defects, graph, boundary_connections):
    best = None
    best_weight = float("inf")
    boundaries = list(boundary_connections.keys())

    def recurse(remaining, matching, weight):
        nonlocal best, best_weight
        if not remaining:
            if weight < best_weight:
                best_weight = weight
                best = matching.copy()
            return
        if weight >= best_weight: 
            return

        defect = remaining[0]
        rest = remaining[1:]

        # Option 1: Match defect to another defect
        for i, other in enumerate(rest):
            remaining_after = rest[:i] + rest[i+1:]
            dist = stabilizer_distance(graph, defect, other)
            if dist == float("inf"): 
                continue
            matching.append((defect, other, dist))
            recurse(remaining_after, matching, weight + dist)
            matching.pop()

        # Option 2: Match defect to a boundary
        for boundary in boundaries:
            dist = boundary_distance(graph, boundary_connections, defect, boundary)
            if dist == float("inf"): 
                continue
            matching.append((defect, boundary, dist))
            recurse(rest, matching, weight + dist)
            matching.pop()

    recurse(defects, [], 0)
    return best

def shortest_path(graph, start, goal):
    queue = deque([(start, [], 0)])
    visited = {start}
    while queue:
        node, path, distance = queue.popleft()
        if node == goal: 
            return path
        for neighbor, qubit in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [qubit], distance + 1))
    return []

def shortest_path_to_boundary(graph, boundary_connections, start, boundary):
    best_path = None
    best_dist = float("inf")
    for stabilizer, boundary_qubits in boundary_connections[boundary].items():
        path = shortest_path(graph, start, stabilizer)
        dist = len(path)
        for q in boundary_qubits:
            if dist + 1 < best_dist:
                best_path = path + [q]
                best_dist = dist + 1
    return best_path if best_path is not None else []

def matching_to_correction(matching, graph, boundary_connections):
    correction = []
    for a, b, _ in matching:
        if not b.startswith("B_"):
            path = shortest_path(graph, a, b)
        else:
            path = shortest_path_to_boundary(graph, boundary_connections, a, b)
        correction.extend(path)
    return sorted(list(set(correction)))

# ============================================================
# 3. Create Full Qiskit Circuit (9 Data Qubits, 8 Ancillas)
# ============================================================

def create_full_surface_code_circuit(errors=None):
    """
    Data Qubits: 0 to 8
    Z-Ancillas:  9 to 12  (Measures physical X errors)
    X-Ancillas:  13 to 16 (Measures physical Z errors)
    """
    qc = QuantumCircuit(17, 8)
    
    # --- A. Inject Arbitrary Physical Errors (Noise Simulation) ---
    if errors:
        for qubit, error_type in errors.items():
            if error_type == "X": qc.x(qubit)
            elif error_type == "Z": qc.z(qubit)
            elif error_type == "Y": qc.y(qubit)
    qc.barrier()
    
    # --- B. Measure Z-Stabilizers (Bit Flips) ---
    for idx, qubits in z_stabilizers_map.items():
        for q in qubits:
            qc.cx(q, 9 + idx)
            
    # --- C. Measure X-Stabilizers (Phase Flips) ---
    for idx, qubits in x_stabilizers_map.items():
        qc.h(13 + idx)
        for q in qubits:
            qc.cx(13 + idx, q)  # Control from Ancilla to target Data
        qc.h(13 + idx)
        
    qc.barrier()
    
    # --- D. Measure registers into classical bits ---
    # Register Layout: [X3, X2, X1, X0, Z3, Z2, Z1, Z0]
    for i in range(4): 
        qc.measure(9 + i, i)       # Z-Ancillas map to clbits 0-3
    for i in range(4): 
        qc.measure(13 + i, 4 + i)  # X-Ancillas map to clbits 4-7
    
    return qc

# ============================================================
# 4. Pipeline Execution Logic
# ============================================================

def process_arbitrary_errors(errors):
    print(f"\n================ Injected Noise Profile: {errors} ================")
    
    # Compile and simulate circuit
    qc = create_full_surface_code_circuit(errors)
    counts = AerSimulator().run(qc, shots=1).result().get_counts()
    
    # Reverse string order because Qiskit clbit 0 sits at the rightmost position
    bitstring = list(counts.keys())[0][::-1]
    
    # Separate sub-bitstrings
    z_bits = bitstring[0:4]  # Clbits 0-3 track bit flips
    x_bits = bitstring[4:8]  # Clbits 4-7 track phase flips
    
    z_defects = [f"Z{i}" for i, bit in enumerate(z_bits) if bit == '1']
    x_defects = [f"X{i}" for i, bit in enumerate(x_bits) if bit == '1']
    
    print(f"Z-Stabilizers Syndrome Readout : {z_bits} -> Defects: {z_defects}")
    print(f"X-Stabilizers Syndrome Readout : {x_bits} -> Defects: {x_defects}")
    
    # Execute independent matchers
    x_correction = []
    if z_defects:
        match_z = boundary_aware_matching(z_defects, graph_z, boundary_connections_z)
        if match_z:
            x_correction = matching_to_correction(match_z, graph_z, boundary_connections_z)
        
    z_correction = []
    if x_defects:
        match_x = boundary_aware_matching(x_defects, graph_x, boundary_connections_x)
        if match_x:
            z_correction = matching_to_correction(match_x, graph_x, boundary_connections_x)
        
    print(f"Calculated X-Correction needed  : Apply Pauli X on Qubits {x_correction}")
    print(f"Calculated Z-Correction needed  : Apply Pauli Z on Qubits {z_correction}")

# ============================================================
# 5. Pipeline Verification Tests
# ============================================================

# Test 1: Combined independent errors (Bit flip on qubit 5, Phase flip on qubit 1)
process_arbitrary_errors({5: "X", 1: "Z"})

# Test 2: A single complex Y error (Will trigger defects in both syndrome graphs at the same time)
process_arbitrary_errors({4: "Y"})
```

```python
import numpy as np
from collections import deque
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# ============================================================
# 1. Complete Unified Decoder Graphs Configuration (Imperative Fix)
# ============================================================

# Define Z Stabilizers
z_stabilizers_map = {}
z_stabilizers_map[0] = [5, 6]
z_stabilizers_map[1] = [0, 1, 4, 5]
z_stabilizers_map[2] = [3, 4, 7, 8]
z_stabilizers_map[3] = [2, 3]

# Define X Stabilizers
x_stabilizers_map = {}
x_stabilizers_map[0] = [0, 1]
x_stabilizers_map[1] = [1, 2, 3, 4]
x_stabilizers_map[2] = [4, 5, 6, 7]
x_stabilizers_map[3] = [7, 8]

# GRAPH A: Graph used for decoding X errors (Bit Flips) via Z-Stabilizers
graph_z = {
    "Z0": [("Z1", 5), ("B_bottom", 6)],
    "Z1": [("B_top", 0), ("B_top", 1), ("Z2", 4), ("Z0", 5)],
    "Z2": [("Z3", 3), ("Z1", 4), ("B_bottom", 7), ("B_bottom", 8)],
    "Z3": [("B_top", 2), ("Z2", 3)],
    "B_top": [("Z1", 0), ("Z1", 1), ("Z3", 2)],
    "B_bottom": [("Z0", 6), ("Z2", 7), ("Z2", 8)],
}

boundary_connections_z = {}
boundary_connections_z["B_top"] = {}
boundary_connections_z["B_top"]["Z1"] = [0, 1]
boundary_connections_z["B_top"]["Z3"] = [2]
boundary_connections_z["B_bottom"] = {}
boundary_connections_z["B_bottom"]["Z0"] = [6]
boundary_connections_z["B_bottom"]["Z2"] = [7, 8]

# GRAPH B: Graph used for decoding Z errors (Phase Flips) via X-Stabilizers
graph_x = {
    "X0": [("X1", 1), ("B_left", 0)],
    "X1": [("X0", 1), ("X2", 4), ("B_right", 2), ("X3", 3)],
    "X2": [("X1", 4), ("X3", 7), ("B_left", 5), ("B_left", 6)],
    "X3": [("X1", 3), ("X2", 7), ("B_right", 8)],
    "B_left": [("X0", 0), ("X2", 5), ("X2", 6)],
    "B_right": [("X1", 2), ("X3", 8)]
}

boundary_connections_x = {}
boundary_connections_x["B_left"] = {}
boundary_connections_x["B_left"]["X0"] = [0]
boundary_connections_x["B_left"]["X2"] = [5, 6]
boundary_connections_x["B_right"] = {}
boundary_connections_x["B_right"]["X1"] = [2]
boundary_connections_x["B_right"]["X3"] = [8]

# ============================================================
# 2. Graph Search Helper Methods
# ============================================================

def stabilizer_distance(graph, start, goal):
    queue = deque([(start, 0)])
    visited = {start}
    while queue:
        node, distance = queue.popleft()
        if node == goal: return distance
        for neighbor, _ in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, distance + 1))
    return float("inf")

def boundary_distance(graph, boundary_connections, start, boundary):
    if boundary not in boundary_connections: return float("inf")
    best = float("inf")
    for stabilizer in boundary_connections[boundary]:
        dist = stabilizer_distance(graph, start, stabilizer)
        if dist != float("inf"):
            best = min(best, dist + 1)
    return best

def boundary_aware_matching(defects, graph, boundary_connections):
    best = None
    best_weight = float("inf")
    boundaries = list(boundary_connections.keys())

    def recurse(remaining, matching, weight):
        nonlocal best, best_weight
        if not remaining:
            if weight < best_weight:
                best_weight = weight
                best = matching.copy()
            return
        if weight >= best_weight: return

        defect = remaining[0]
        rest = remaining[1:]

        for i, other in enumerate(rest):
            remaining_after = rest[:i] + rest[i+1:]
            dist = stabilizer_distance(graph, defect, other)
            if dist == float("inf"): continue
            matching.append((defect, other, dist))
            recurse(remaining_after, matching, weight + dist)
            matching.pop()

        for boundary in boundaries:
            dist = boundary_distance(graph, boundary_connections, defect, boundary)
            if dist == float("inf"): continue
            matching.append((defect, boundary, dist))
            recurse(rest, matching, weight + dist)
            matching.pop()

    recurse(defects, [], 0)
    return best

def shortest_path(graph, start, goal):
    queue = deque([(start, [], 0)])
    visited = {start}
    while queue:
        node, path, distance = queue.popleft()
        if node == goal: return path
        for neighbor, qubit in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [qubit], distance + 1))
    return []

def shortest_path_to_boundary(graph, boundary_connections, start, boundary):
    best_path = None
    best_dist = float("inf")
    for stabilizer, boundary_qubits in boundary_connections[boundary].items():
        path = shortest_path(graph, start, stabilizer)
        dist = len(path)
        for q in boundary_qubits:
            if dist + 1 < best_dist:
                best_path = path + [q]
                best_dist = dist + 1
    return best_path if best_path is not None else []

def matching_to_correction(matching, graph, boundary_connections):
    correction = []
    for a, b, _ in matching:
        if not b.startswith("B_"):
            path = shortest_path(graph, a, b)
        else:
            path = shortest_path_to_boundary(graph, boundary_connections, a, b)
        correction.extend(path)
    return sorted(list(set(correction)))

# ============================================================
# 3. Create Full Two-Round Qiskit Circuit
# ============================================================

def create_two_round_circuit(errors=None):
    qc = QuantumCircuit(17, 16) 
    
    # --- ROUND 1: Establish baseline configuration ---
    for idx, qubits in z_stabilizers_map.items():
        for q in qubits: qc.cx(q, 9 + idx)
    for idx, qubits in x_stabilizers_map.items():
        qc.h(13 + idx)
        for q in qubits: qc.cx(13 + idx, q)
        qc.h(13 + idx)
        
    for i in range(4): qc.measure(9 + i, i)       
    for i in range(4): qc.measure(13 + i, 4 + i)  
    
    for i in range(9, 17): qc.reset(i)
    qc.barrier()
    
    # --- NOISE INJECTION SYSTEM ---
    if errors:
        for qubit, error_type in errors.items():
            if error_type == "X": qc.x(qubit)
            elif error_type == "Z": qc.z(qubit)
            elif error_type == "Y": qc.y(qubit)
    qc.barrier()
    
    # --- ROUND 2: Post-Fault Syndrome Extraction ---
    for idx, qubits in z_stabilizers_map.items():
        for q in qubits: qc.cx(q, 9 + idx)
    for idx, qubits in x_stabilizers_map.items():
        qc.h(13 + idx)
        for q in qubits: qc.cx(13 + idx, q)
        qc.h(13 + idx)
        
    for i in range(4): qc.measure(9 + i, 8 + i)    
    for i in range(4): qc.measure(13 + i, 12 + i)  
    
    return qc

# ============================================================
# 4. Pipeline Execution Logic
# ============================================================

def process_arbitrary_errors(errors):
    print(f"\n================ Injected Noise Profile: {errors} ================")
    
    qc = create_two_round_circuit(errors)
    counts = AerSimulator().run(qc, shots=1).result().get_counts()
    bitstring = list(counts.keys())[0][::-1] 
    
    round1_z = [int(b) for b in bitstring[0:4]]
    round1_x = [int(b) for b in bitstring[4:8]]
    round2_z = [int(b) for b in bitstring[8:12]]
    round2_x = [int(b) for b in bitstring[12:16]]
    
    # XOR evaluation isolates the fault changes
    delta_z = [str(r1 ^ r2) for r1, r2 in zip(round1_z, round2_z)]
    delta_x = [str(r1 ^ r2) for r1, r2 in zip(round1_x, round2_x)]
    
    z_defects = [f"Z{i}" for i, bit in enumerate(delta_z) if bit == '1']
    x_defects = [f"X{i}" for i, bit in enumerate(delta_x) if bit == '1']
    
    print(f"Z-Stabilizers Delta Syndrome   : {''.join(delta_z)} -> Defects: {z_defects}")
    print(f"X-Stabilizers Delta Syndrome   : {''.join(delta_x)} -> Defects: {x_defects}")
    
    x_correction, z_correction = [], []
    if z_defects:
        match_z = boundary_aware_matching(z_defects, graph_z, boundary_connections_z)
        if match_z: x_correction = matching_to_correction(match_z, graph_z, boundary_connections_z)
    if x_defects:
        match_x = boundary_aware_matching(x_defects, graph_x, boundary_connections_x)
        if match_x: z_correction = matching_to_correction(match_x, graph_x, boundary_connections_x)
        
    print(f"Calculated X-Correction needed  : Apply Pauli X on Qubits {x_correction}")
    print(f"Calculated Z-Correction needed  : Apply Pauli Z on Qubits {z_correction}")

# ============================================================
# 5. Execution Verification
# ============================================================
process_arbitrary_errors({5: "X", 1: "Z"})
process_arbitrary_errors({4: "Y"})
```
